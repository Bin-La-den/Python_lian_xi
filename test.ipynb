```python
import numpy as np
import matplotlib.pyplot as plt

# 解决 matplotlib 中文显示为方块的问题（系统使用思源黑体 Source Han Sans CN）
plt.rcParams['font.sans-serif'] = ['Source Han Sans CN', 'Droid Sans Fallback', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False  # 解决坐标轴负号 '−' 显示为方块的问题
```

```python
# 生成模拟数据：在正弦曲线基础上加入一些随机噪声
np.random.seed(42)
X = np.linspace(0, 10, 20)
y_true = np.sin(X)                     # 真实的潜在规律（我们不知道）
y_noise = np.random.randn(20) * 0.3   # 随机噪声
y = y_true + y_noise                  # 我们实际观测到的数据
```

```python
plt.scatter(X, y, label='观测数据 (含噪声)', color='blue', alpha=0.6)
plt.plot(X, y_true, label='真实规律 (y=sin(x))', color='green', linewidth=2)
plt.xlabel('X')
plt.ylabel('y')
plt.title('数据与潜在规律')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure()
plt.plot(X, y_noise, marker='o', color='red', linewidth=1, label='噪声 y_noise')
plt.axhline(0, color='black', linewidth=0.8)                 # 0 参考线（声围绕它波动）
plt.axhline(0.3, color='gray', linestyle='--', alpha=0.7)    # +1σ 参考线
plt.axhline(-0.3, color='gray', linestyle='--', alpha=0.7)   # -1σ 参考线
plt.xlabel('X')
plt.ylabel('y_noise')
plt.title('随机噪声 y_noise')
plt.legend()
plt.grid(True)
plt.show()
```
